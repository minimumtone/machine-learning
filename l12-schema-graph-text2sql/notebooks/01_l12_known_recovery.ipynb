# 01 - Known L1₂ Compound RecoveryVerify that known L1₂ compounds can be retrieved from the database.

```python
import syssys.path.insert(0, '..')from llm.sql_generator import pipelinefrom safety.sql_guard import execute_sql
```

```python
query = "L1₂構造を持つ化合物を一覧にして"result = pipeline(query)print("Generated SQL:")print(result["sql"])
```

```python
exec_result = execute_sql(result["sql"])if exec_result["success"]:    import pandas as pd    df = pd.DataFrame(exec_result["rows"], columns=exec_result["columns"])    print(f"Found {len(df)} L1₂ compounds")    display(df)else:    print("Execution failed:", exec_result["errors"])
```

```python
known = ["Ni3Al", "Ni3Ga", "Ni3Ge", "Co3Ti", "Al3Sc", "Al3Ti", "Pt3Al", "Ir3Nb"]if exec_result["success"]:    found = set(df["formula"].tolist()) if "formula" in df.columns else set()    recovered = [k for k in known if k in found]    missing = [k for k in known if k not in found]    print(f"Recovery rate: {len(recovered)}/{len(known)} = {len(recovered)/len(known):.0%}")    print(f"Recovered: {recovered}")    print(f"Missing: {missing}")
```

```python
# Save resultsif exec_result["success"]:    df.to_csv("../known_l12_recovery.csv", index=False)    print("Saved to known_l12_recovery.csv")
```
